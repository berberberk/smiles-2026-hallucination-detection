```python
from pathlib import Path
import shutil
import subprocess

def run(cmd, cwd=None):
    print(f'$ {cmd}')
    subprocess.run(cmd, shell=True, check=True, cwd=cwd)
```

```python
REPO_URL = 'https://github.com/berberberk/smiles-2026-hallucination-detection'
REPO_PATH = Path('/content/smiles-2026-hallucination-detection')

if not REPO_PATH.exists():
    run(f'git clone {REPO_URL} {REPO_PATH}')

run('pip install -q -r requirements.txt', cwd=REPO_PATH)
```

```python
run('python solution.py', cwd=REPO_PATH)
```

```python
from google.colab import drive

drive.mount('/content/drive')
OUT = Path('/content/drive/MyDrive/smiles_2026_outputs')
OUT.mkdir(parents=True, exist_ok=True)

for name in ['results.json', 'predictions.csv']:
    shutil.copy2(REPO_PATH / name, OUT / name)

print(OUT)
```
